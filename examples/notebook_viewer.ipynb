# Notebook viewer

Use a local Python kernel with `mathfirst[viz]` installed. From this checkout, run `uv sync --locked --extra viz`, then select the `.venv` environment in your notebook editor. The kernel also needs `ipykernel`; let your editor install it if needed.

Run the cells in order. The browser opens on the same computer as Python. Keep the kernel running while you interact, and run the final shutdown cell when finished. Before rerunning construction cells, close the existing viewer.


```python
from mathfirst import Polynomial, Variable
from mathfirst.viz import FunctionGraph, Viewer
```

## Define the mathematics

The polynomial owns its mathematical structure. Evaluating it is independent of visualization.


```python
x = Variable("x")
p = Polynomial(x**3 - 2 * x + 1, x)
p(2)  # Exact value: 5.
```

## Describe a visual interpretation

`FunctionGraph` says to represent the polynomial as a graph. It does not start a server or compute samples.


```python
graph = FunctionGraph(p)
```

## Construct the viewer

Construction does not start the runtime. Close the existing viewer before rerunning this cell so its server does not remain in the background.


```python
viewer = Viewer(graph)
```

## Start explicitly in background mode

`show()` blocks by default in both scripts and notebooks. `block=False` returns once the server is ready, so you can continue using the notebook. Repeating this startup cell reuses the running server.


```python
viewer.show(block=False)
```

Inspect the running URL. If you prefer to open it manually, use `viewer.show(block=False, open_browser=False)` in the startup cell. In the browser, scroll to zoom, drag to pan, and choose **Reset view** to restore the initial range.


```python
viewer.url
```

## Stop the server

Closing the browser tab leaves the Python server running. Call `close()` explicitly when finished; repeated calls are safe.


```python
viewer.close()
```

`viewer.url` is now `None`. You can rerun the startup cell to start the same viewer again. If you change the mathematics, close this viewer before reconstructing the realization and viewer.
